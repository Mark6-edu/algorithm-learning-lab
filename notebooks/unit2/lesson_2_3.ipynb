## 실습 1. 순차 구조

두 정수를 입력받아 합계를 출력해 봅시다.

```python
num1 = int(input("첫 번째 정수: "))
num2 = int(input("두 번째 정수: "))

result = num1 + num2

print("합계:", result)
```

### 확인 문제
- 두 수의 **합**이 아니라 **차**를 출력하도록 코드를 수정해 보세요.
- 세 개의 정수를 입력받아 합계를 출력하도록 확장해 보세요.

## 실습 2. 선택 구조

점수를 입력받아 60점 이상이면 `합격`, 그렇지 않으면 `불합격`을 출력해 봅시다.

```python
score = int(input("점수: "))

if score >= 60:
    print("합격")
else:
    print("불합격")
```

### 확인 문제
- 90점 이상 `A`, 80점 이상 `B`, 70점 이상 `C`, 그 외 `D`가 출력되도록 수정해 보세요.

## 실습 3. 반복 구조

1부터 10까지 차례대로 출력해 봅시다.

```python
for number in range(1, 11):
    print(number)
```

### 확인 문제
- 1부터 20까지 출력하도록 수정해 보세요.
- 1부터 20까지의 **짝수만** 출력하도록 수정해 보세요.

## 실습 4. 도전 문제

정수 `n`을 입력받아 1부터 `n`까지의 합을 구해 봅시다.

```python
n = int(input("정수 n: "))

total = 0

for number in range(1, n + 1):
    total += number

print("1부터", n, "까지의 합:", total)
```

## 마무리

이번 실습에서 사용한 구조를 정리해 봅시다.

- **순차 구조**: 명령을 정해진 순서대로 실행
- **선택 구조**: 조건에 따라 서로 다른 명령 실행
- **반복 구조**: 같은 작업을 여러 번 반복
