# 2-4 알고리즘 성능 분석 Python 실습

이번 실습에서는 **입력 크기, 수행 횟수, O(n), O(n²), 실행 시간 비교**를 직접 확인합니다.

코드를 실행한 뒤 입력 크기 `n`을 바꾸면서 결과가 어떻게 달라지는지 관찰해 보세요.

## 실습 1. O(n) 수행 횟수 확인

리스트의 모든 원소를 한 번씩 확인하는 알고리즘입니다.

```python
numbers = [10, 20, 30, 40, 50]

count = 0

for number in numbers:
    count += 1

print("데이터 개수:", len(numbers))
print("수행 횟수:", count)
```

### 확인 문제
- 데이터가 10개가 되도록 리스트를 수정해 보세요.
- 수행 횟수는 어떻게 변하는지 확인해 보세요.

## 실습 2. O(n²) 수행 횟수 확인

이중 반복문에서는 입력 크기가 커질수록 수행 횟수가 훨씬 빠르게 증가합니다.

```python
numbers = [1, 2, 3, 4, 5]

count = 0

for i in numbers:
    for j in numbers:
        count += 1

print("데이터 개수:", len(numbers))
print("수행 횟수:", count)
```

### 확인 문제
- 데이터 개수를 10개로 늘려 보세요.
- O(n)과 O(n²)의 수행 횟수가 얼마나 차이 나는지 비교해 보세요.

## 실습 3. 입력 크기에 따른 증가 비교

`n`이 증가할 때 O(1), O(n), O(n²)의 작업량을 표로 확인해 봅시다.

```python
sizes = [10, 100, 1000]

print("n\tO(1)\tO(n)\tO(n²)")

for n in sizes:
    print(f"{n}\t{1}\t{n}\t{n**2}")
```

## 실습 4. 실제 실행 시간 비교

아래 코드는 선형 반복과 이중 반복의 실행 시간을 직접 측정합니다.

> 실행 환경에 따라 시간 값 자체는 달라질 수 있으므로, **증가하는 경향**을 관찰하는 것이 중요합니다.

```python
import time

n = 2000

start = time.perf_counter()
for i in range(n):
    pass
linear_time = time.perf_counter() - start

start = time.perf_counter()
for i in range(n):
    for j in range(n):
        pass
quadratic_time = time.perf_counter() - start

print("O(n) 실행 시간:", linear_time)
print("O(n²) 실행 시간:", quadratic_time)
```

### 도전 문제
- `n = 500`, `1000`, `2000`으로 바꾸어 실행 시간을 비교해 보세요.
- 입력 크기가 두 배가 되었을 때 O(n)과 O(n²)의 실행 시간이 어떻게 달라지는지 관찰해 보세요.

## 마무리

- **O(1)**: 입력 크기와 관계없이 작업량이 거의 일정
- **O(n)**: 입력 크기에 비례하여 작업량 증가
- **O(n²)**: 입력 크기의 제곱에 비례하여 작업량 증가

Big-O 표기법은 실제 초 단위 시간을 뜻하는 것이 아니라, **입력 크기가 커질 때 필요한 연산량이 얼마나 빠르게 증가하는지**를 나타냅니다.
