## 실습 1. 문제를 작은 작업으로 나누기

교실 온도를 확인하여 기준 온도 이상이면 선풍기를 켜는 문제를 생각해 봅시다.

문제를 다음과 같이 나눌 수 있습니다.
1. 현재 온도를 확인한다.
2. 기준 온도를 정한다.
3. 두 온도를 비교한다.
4. 선풍기 상태를 결정한다.

```python
current_temp = 29
standard_temp = 28

print("현재 온도:", current_temp)
print("기준 온도:", standard_temp)
```

## 실습 2. 추상화하기

문제 해결에 필요한 정보만 남겨 봅시다.

- 필요한 정보: 현재 온도, 기준 온도, 선풍기 상태
- 필요하지 않은 정보: 교실 벽 색상, 학생 이름 등

```python
current_temp = 29
standard_temp = 28

if current_temp >= standard_temp:
    fan = "ON"
else:
    fan = "OFF"

print("선풍기 상태:", fan)
```

## 실습 3. 모델을 데이터 구조로 표현하기

도서관 책 정보를 간단한 Python 딕셔너리로 표현해 봅시다.

```python
book = {
    "title": "알고리즘 설계",
    "available": True,
    "location": "A-03"
}

print("책 제목:", book["title"])
print("대출 가능:", book["available"])
print("위치:", book["location"])
```

## 실습 4. 도전 문제

책 제목을 입력받아 대출 가능 여부를 알려주는 간단한 프로그램을 만들어 봅시다.

```python
books = {
    "알고리즘 설계": True,
    "자료 구조": False,
    "Python 기초": True
}

title = input("책 제목을 입력하세요: ")

if title in books:
    if books[title]:
        print("대출 가능합니다.")
    else:
        print("현재 대출 중입니다.")
else:
    print("도서 목록에 없습니다.")
```
