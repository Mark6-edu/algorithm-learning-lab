## 실습 1. 현재 상태와 목표 상태 확인

학생들의 점수가 주어졌을 때 가장 높은 점수를 찾는 문제를 살펴봅시다.

- 현재 상태: 여러 점수가 있지만 최댓값을 모름
- 목표 상태: 가장 높은 점수를 알아냄

```python
scores = [78, 92, 85, 67, 95]

print("현재 점수 목록:", scores)
print("목표: 가장 높은 점수 찾기")
```

## 실습 2. 필요한 작업을 Python으로 표현

점수를 하나씩 비교하여 최댓값을 찾습니다.

```python
scores = [78, 92, 85, 67, 95]

max_score = scores[0]

for score in scores:
    if score > max_score:
        max_score = score

print("가장 높은 점수:", max_score)
```

## 실습 3. 입력 → 처리 → 출력으로 분석

두 수 중 큰 값을 찾는 문제를 입력, 처리, 출력으로 나누어 봅시다.

```python
a = int(input("첫 번째 수: "))
b = int(input("두 번째 수: "))

if a > b:
    result = a
else:
    result = b

print("더 큰 수:", result)
```
